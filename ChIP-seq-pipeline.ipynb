```python
%%bash
## check for regions that behaviour as blacklist. Each bioproject has its own blacklist
## for single-end ChIP-seq
./scripts/ChIP-seq/Script-04-macs3-blacklist.sh . SRR2530361 PRJNA236156 single
# input samples
./scripts/ChIP-seq/Script-04-macs3-blacklist.sh . SRR2530363 PRJNA236156 single
    
cd intermediate/04.macs3_blacklist_PRJNA236156
cat */*broadPeak | awk '{if ($NF > 10) print $0}' | bedtools sort -i - | bedtools merge -delim "|"  -c 4 -o count,collapse -i - | awk '{if ($4 > 1) print $0}' > black.list.bed
    

## for paired-end ChIP-seq
./scripts/ChIP-seq/Script-04-macs3-blacklist.sh . SRR10961892 PRJNA602708 paired
# input samples
./scripts/ChIP-seq/Script-04-macs3-blacklist.sh . SRR10961877 PRJNA602708 paired
    
cd intermediate/04.macs3_blacklist_PRJNA602708
cat */*broadPeak | awk '{if ($NF > 10) print $0}' | bedtools sort -i - | bedtools merge -delim "|"  -c 4 -o count,collapse -i - | awk '{if ($4 > 1) print $0}' > black.list.bed
```
